```python
from pathlib import Path

data_dir = Path("data/trashnet")
counts = {d.name: len(list(d.glob("*.jpg"))) for d in sorted(data_dir.iterdir()) if d.is_dir()}
print(counts)
```

```python
import matplotlib.pyplot as plt
from PIL import Image

fig, axes = plt.subplots(2, 3, figsize=(12, 8))
for ax, class_dir in zip(axes.flat, sorted(data_dir.iterdir())):
    img = Image.open(next(class_dir.glob("*.jpg")))
    ax.imshow(img)
    ax.set_title(f"{class_dir.name} {img.size}")
    ax.axis("off")
plt.show()
```
